```python
import hashlib, importlib.metadata, json, os, subprocess, sys
from pathlib import Path
os.environ['EQUINEXUS_SOURCE_COMMIT'] = SOURCE_COMMIT
os.environ['EQUINEXUS_SOURCE_BUNDLE_SHA256'] = SOURCE_BUNDLE_SHA256
os.environ['EQUINEXUS_EXTERNAL_SUPERVISED_DATA'] = 'true'
os.environ['EQUINEXUS_TRAINING_DATA_DESCRIPTION'] = 'SWEFil CC-BY-4.0 supervised pretraining, followed by fine-tuning on competition-provided training annotations only'
os.environ['EQUINEXUS_PRETRAINED_SOURCE'] = 'Ultralytics YOLO11-seg initialized from COCO weights and supervised-pretrained on audited SWEFil CC-BY-4.0 masks'
os.environ['EQUINEXUS_PRETRAINED_LICENSE'] = 'AGPL-3.0 model/code; SWEFil CC-BY-4.0 attribution retained in external-data provenance'
os.environ['YOLO_CONFIG_DIR'] = '/kaggle/working/equinexus/yolo-config'
os.environ['TORCH_HOME'] = '/kaggle/working/equinexus/torch-home'
Path(os.environ['YOLO_CONFIG_DIR']).mkdir(parents=True, exist_ok=True)
Path(os.environ['TORCH_HOME']).mkdir(parents=True, exist_ok=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check', '--no-cache-dir', '-r', str(REPOSITORY_PATH / 'requirements-detector.txt')], check=True)
import torch, ultralytics
devices = [{'name': torch.cuda.get_device_name(i), 'total_memory_bytes': torch.cuda.get_device_properties(i).total_memory} for i in range(torch.cuda.device_count())]
print({'python': sys.version.split()[0], 'torch': torch.__version__, 'cuda': torch.version.cuda, 'ultralytics': ultralytics.__version__, 'cuda_available': torch.cuda.is_available(), 'devices': devices})
assert torch.cuda.is_available(), 'A Kaggle GPU is required for this run'
INPUT_ROOT = Path('/kaggle/input')
RUN_SLUG = 'yolo11s-seg-swefil-30ep'
EXPECTED_CHECKPOINT_SHA256 = '6e1eedabd75a98783d4e840be7b98f399f0930197886bba560758e7d0dbb1bbd'
EXPECTED_EXTERNAL_MANIFEST_SHA256 = '0a8f6e7f313d78bfff4935645e919e9825c17233e8dfa880cfae3f69628d3d8d'
EXPECTED_HOLDOUT_SHA256 = '30cb2ec52437daf97e769a3a61bf83350062e6cf4f5d00af1f8399649065e48e'
checkpoint_candidates = [p for p in INPUT_ROOT.rglob('best.pt') if p.parent.name == RUN_SLUG]
if len(checkpoint_candidates) != 1:
    raise FileNotFoundError(f'Expected one attached SWEFil checkpoint, found: {checkpoint_candidates}')
checkpoint_path = checkpoint_candidates[0]
sha = hashlib.sha256(checkpoint_path.read_bytes()).hexdigest()
assert sha == EXPECTED_CHECKPOINT_SHA256, f'Unexpected SWEFil checkpoint hash: {sha}'
pretrain_root = checkpoint_path.parent
pretrain_metadata = json.loads((pretrain_root / 'run-metadata.json').read_text())
external_provenance = json.loads((pretrain_root / 'external-data-provenance.json').read_text())
assert pretrain_metadata['status'] == 'complete'
assert pretrain_metadata['frozen_holdout_used_for_training_or_selection'] is False
assert pretrain_metadata['frozen_holdout_sha256'] == EXPECTED_HOLDOUT_SHA256
assert external_provenance['included_manifest_sha256'] == EXPECTED_EXTERNAL_MANIFEST_SHA256
assert external_provenance['oof_safe_date_exclusion'] is True
print({'checkpoint': str(checkpoint_path), 'checkpoint_sha256': sha, 'external_manifest_sha256': external_provenance['included_manifest_sha256']})
```

```python
import shutil
run_root = Path('/kaggle/working/equinexus/yolo11s-seg-swefil-ft-25ep')
run_root.mkdir(parents=True, exist_ok=True)
shutil.copy2(pretrain_root / 'external-data-provenance.json', run_root / 'pretrained-external-data-provenance.json')
shutil.copy2(pretrain_root / 'run-metadata.json', run_root / 'swefil-pretraining-run-metadata.json')
command = [sys.executable, '-u', str(REPOSITORY_PATH / 'scripts' / 'run_yolo_detector_experiment.py'), '--repo-root', str(REPOSITORY_PATH), '--run-name', 'yolo11s-seg-swefil-ft-25ep', '--model', str(checkpoint_path), '--image-size', '1024', '--epochs', '25', '--batch-size', '4', '--workers', '4', '--patience', '8', '--device', '0']
print('Starting SWEFil warm-start fine-tuning:', ' '.join(command), flush=True)
subprocess.run(command, check=True)
```
